```python
import polars as pl
```

```python
df = pl.DataFrame(
    {
        "first_name": ["dan", "stan", "ron", "dawn"],
        "last_name": ["flanson", "dfs", "thoman", "ronin"]
    }
)
df
```

```python
df = pl.from_dicts(
    [
        {"first_name": "dan", "second_name": "flansons"},
        {"first_name": "van", "second_name": "sldfj"},
        {"first_name": "man", "second_name": "stuary"},
        {"first_name": "rom", "second_name": "parker"},
    ]
)
df
```

```python
df = pl.from_records(
    [
        ["dan", "flanson"],
        ["van", "robert"],
        ["gan", "xavier"],
        ["xan", "robert"],
    ],
    schema=["first_name", "last_name"],
)
df
```

```python
yellow_rides_column_rename_mapping = {
    "VendorID": "vendor_id",
    "RatecodeID": "rate_code_id",
    "PULocationID": "pickup_location_id",
    "DOLocationID": "dropoff_location_id",
    "Airport_fee": "airport_fee",
}

march_yellow_rides_df = (
    pl.read_parquet("./../data/yellow_tripdata_2024-03.parquet").rename(
        yellow_rides_column_rename_mapping
    ).with_columns(
        pl.col("tpep_pickup_datetime").dt.date().alias("tpep_pickup_date")
    )
)
```

```python
march_yellow_rides_df
```

```python
march_yellow_rides_df.write_parquet(
    "./../data/parquet_partitions/partitioned_march_rides/",
    use_pyarrow=True,
    pyarrow_options={"partition_cols": ["tpep_pickup_date"]}
)
```

```python
loaded_df_partition = pl.read_parquet(
    "./../data/parquet_partitions/partitioned_march_rides/",
    use_pyarrow=True,
)
loaded_df_partition
```

```python
loaded_df_partition = pl.read_parquet(
    "./../data/parquet_partitions/partitioned_march_rides/tpep_pickup_date=2024-03-06/",
    use_pyarrow=True,
)
loaded_df_partition
```
